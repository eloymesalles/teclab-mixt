# Teclab: resum d'un text via la passarel·la

Aquest notebook crida la passarel·la LiteLLM de la plataforma (API compatible amb OpenAI; adreça a `LITELLM_BASE_URL`) amb la clau de l'espai de treball (`LITELLM_KEY`) i demana un resum al model de l'àlies `LITELLM_MODEL` (`chat-default`).

El mateix codi funciona a **Coder** i a **JupyterHub**, a la demo i al sandbox del Teclab: la imatge és la mateixa i no porta cap adreça cuita; les variables les posa la plataforma.

```python
import os
from openai import OpenAI

base_url = os.environ.get("LITELLM_BASE_URL", "").strip()
key = os.environ.get("LITELLM_KEY", "").strip()
model = os.environ.get("LITELLM_MODEL", "").strip() or "chat-default"
for nom, valor in (("LITELLM_BASE_URL", base_url), ("LITELLM_KEY", key)):
    if not valor:
        raise SystemExit(f"Falta la variable d'entorn {nom}: la posa la plataforma a l'espai (vegeu .env.example)")
client = OpenAI(base_url=base_url, api_key=key)
print("passarel·la:", base_url, "· model:", model)
```

```python
text = """El Teclab és el laboratori d'intel·ligència artificial de la Universitat Jaume I.
Està avaluant dos entorns d'escriptori gestionats, Coder i JupyterHub, per oferir espais
de treball per projecte des del navegador, amb VS Code i notebooks. Els dos entorns
comparteixen la mateixa imatge i accedeixen als models a través d'una passarel·la comuna."""

resposta = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "system", "content": "Resumeix en català, en dues frases com a màxim."},
        {"role": "user", "content": text},
    ],
    max_tokens=1500,  # els models de raonament gasten tokens pensant
    temperature=0.3,
)
print(resposta.choices[0].message.content)
print("\nmodel:", resposta.model, "· tokens:", resposta.usage.total_tokens if resposta.usage else "?")
```
